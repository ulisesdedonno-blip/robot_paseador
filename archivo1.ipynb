```python
def sumar(numero1, numero2):
    """Suma dos números e imprime el resultado."""
    resultado = numero1 + numero2
    # Usamos f-strings (cadenas con formato) para mostrar el texto de forma limpia y ordenada
    print(f"\nResultado: {numero1} + {numero2} = {resultado}\n")

def restar(numero1, numero2):
    """Resta dos números e imprime el resultado."""
    resultado = numero1 - numero2
    print(f"\nResultado: {numero1} - {numero2} = {resultado}\n")

def multiplicar(numero1, numero2):
    """Multiplica dos números e imprime el resultado."""
    resultado = numero1 * numero2
    print(f"\nResultado: {numero1} * {numero2} = {resultado}\n")

def dividir(numero1, numero2):
    """Divide dos números validando que el divisor no sea cero para evitar errores."""
    if numero2 != 0:
        resultado = numero1 / numero2
        print(f"\nResultado: {numero1} / {numero2} = {resultado}\n")
    else:
        print("\n❌ Error: El número 2 es igual a 0 y no se puede dividir.\n")

def mostrar_menu():
    """Muestra las opciones disponibles en pantalla."""
    print("====================")
    print("    CALCULADORA     ")
    print("====================")
    print("1. Sumar")
    print("2. Restar")
    print("3. Multiplicar")
    print("4. Dividir")
    print("5. Salir")

def main():
    """Función principal que controla el flujo del programa con un bucle while."""
    # Un bucle 'while True' crea un ciclo infinito que solo se detiene con un 'break'
    while True:
        mostrar_menu()
        opcion = input("Elige una opción (1-5): ")

        # Si el usuario elige salir, rompemos el bucle
        if opcion == "5":
            print("\n¡Gracias por usar la calculadora! Hasta luego.")
            break

        # Verificamos si la opción elegida es válida antes de pedir los números
        if opcion in ("1", "2", "3", "4"):
            try:
                # Solicitamos los números dentro del bucle para poder hacer múltiples operaciones
                numero1 = float(input("Inserta el número 1: "))
                numero2 = float(input("Inserta el número 2: "))
            except ValueError:
                print("\n❌ Por favor, introduce un número válido.\n")
                continue # Vuelve al inicio del bucle si hubo un error al escribir el número

            # Ejecutamos la función correspondiente según la opción del usuario
            if opcion == "1":
                sumar(numero1, numero2)
            elif opcion == "2":
                restar(numero1, numero2)
            elif opcion == "3":
                multiplicar(numero1, numero2)
            elif opcion == "4":
                dividir(numero1, numero2)
        else:
            print("\n❌ Opción no válida. Por favor, elige un número del 1 al 5.\n")

# Punto de entrada de nuestro programa en Python
if __name__ == "__main__":
    main()
```
